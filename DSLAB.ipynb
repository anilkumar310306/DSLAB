```python
X = df[["Annual Income (k$)", "Spending Score (1-100)"]]

print("\n")
print("=" * 70)
print("SELECTED FEATURES")
print("=" * 70)
print(X.head())
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("\n")
print("=" * 70)
print("STANDARDIZED FEATURES")
print("=" * 70)
print(X_scaled[:5])
```

```python
inertia = []  # reset list

for k in range(1, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), inertia, marker="o")
plt.title("Elbow Method for Optimal Number of Clusters")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)
print(df['Cluster'])
```

```python
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
df["Cluster"] = kmeans.fit_predict(X_scaled)

print("\n")
print("=" * 70)
print("CUSTOMER CLUSTERS")
print("=" * 70)
print(df[["CustomerID","Annual Income (k$)","Spending Score (1-100)","Cluster"]].head(20))
```

```python
cluster_centers = scaler.inverse_transform(kmeans.cluster_centers_)
centers_df = pd.DataFrame(cluster_centers, columns=["Annual Income (k$)", "Spending Score (1-100)"])
centers_df.index.name = "Cluster"

print("\n")
print("=" * 70)
print("CLUSTER CENTERS")
print("=" * 70)
print(centers_df)
```

```python
print("\n")
print("=" * 70)
print("CUSTOMER COUNT PER CLUSTER")
print("=" * 70)
print(df["Cluster"].value_counts().sort_index())

summary = df.groupby("Cluster")[["Annual Income (k$)", "Spending Score (1-100)"]].mean()
summary["Customer Count"] = df["Cluster"].value_counts()
summary = summary.sort_index()
```

```python
print("\n")
print("=" * 70)
print("CUSTOMER SEGMENT SUMMARY")
print("=" * 70)
print(summary)
```

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(data=df, x="Annual Income (k$)", y="Spending Score (1-100)", hue="Cluster", palette="tab10", s=100)
plt.scatter(cluster_centers[:, 0], cluster_centers[:, 1], marker="X", s=250, c="black", label="Cluster Centers")
plt.title("Customer Personality Analysis using K-Means", fontsize=15)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig("customer_segments_kmeans.png", dpi=300)
plt.show()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
```

```python
texts = ["I am happy", "I am angry", "I feel sad", "I love this"]
labels = ["happy", "angry", "sad", "happy"]

for sentence, emotion in zip(texts, labels):
    print(f"Text: {sentence} --> Emotion: {emotion}")
```

```python
vec = CountVectorizer()
X = vec.fit_transform(texts)

print("Vocabulary:", vec.get_feature_names_out())
print("Vectorized Data:\n", X.toarray())
```

```python
model = MultinomialNB()
model.fit(X, labels)
print("Model Training Completed Successfully")
```

```python
new_text = ["I am so sad today"]
test = vec.transform(new_text)

print("Vectorized Test Data:\n", test.toarray())
prediction = model.predict(test)
print("Predicted Emotion:", prediction[0])
```

```python
sample_sentences = [
    "I love my friends",
    "I am angry with you",
    "I feel very sad",
    "Today I am happy"
]

for sentence in sample_sentences:
    sample_vector = vec.transform([sentence])
    result = model.predict(sample_vector)
    print(f"{sentence} --> {result[0]}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

texts = ["I am happy", "I am angry", "I feel sad", "I love this"]
labels = ["happy", "angry", "sad", "happy"]

for sentence, emotion in zip(texts, labels):
    print(f"Text: {sentence} --> Emotion: {emotion}")

vec = CountVectorizer()
X = vec.fit_transform(texts)

print("\nVocabulary:", vec.get_feature_names_out())
print("\nVectorized Data:\n", X.toarray())

model = MultinomialNB()
model.fit(X, labels)
print("\nModel Training Completed Successfully")

new_text = ["I am so sad today"]
test = vec.transform(new_text)
print("\nVectorized Test Data:\n", test.toarray())
prediction = model.predict(test)
print("\nPredicted Emotion:", prediction[0])

sample_sentences = [
    "I love my friends",
    "I am angry with you",
    "I feel very sad",
    "Today I am happy"
]

print("\nMultiple Predictions:")
for sentence in sample_sentences:
    sample_vector = vec.transform([sentence])
    result = model.predict(sample_vector)
    print(f"{sentence} --> {result[0]}")

print("\nTEXT EMOTION DETECTION COMPLETED")
```
